
### Image Conversion

```matlab


I1 = imread("images/honey-comb.jpg");
imshow(rgb2gray(I1))
```

```matlab
for i = [2 5 50]
    [X, map] = rgb2ind(I1, i);
    figure; imshow(X, map)
end
```


<center></center>

### Sampling

```matlab
[r, c, s] = size(I1);
N = 2;
I1s = I1/255;
y = zeros(size(I1s));
y(1:N:r, 1:N:c, :) = I1s(1:N:r, 1:N:c, :);
imshow(y)
```


### Noisy Image

```matlab
imshow(imnoise(I1s, "gaussian", 0, 1));
```

```matlab
imshow(imnoise(I1s, "salt & pepper", 0.19));
```


### Bit Resolution

```matlab


b = [7 6 5];
for i = 1 : length(b)
    d = 2^b(i);
    z = round(I1/d);
    figure
    imshow(z*d)
end
```

### Image Operations

```matlab
I2 = imread("images/zebra.jpg");
imshow(I2, []);
```

```matlab


I1c = I1(1:627, 1:1200, :);


imshow(0.3 * I2);
```

```matlab


imshow(0.7 * I2 + 0.3 * I1c);
```

```matlab
imshow(0.7 * I1c + 0.3 * I2);
```

```matlab


% imadd()
imshow(imadd(I2, I1c));
```

```matlab
imshow(imsubtract(I2, I1c));
```

```matlab
imshow(I2 - I1c);
```

```matlab
imshow(I1c - I2);
imshow(imabsdiff(I2, I1c));
%image subtraction is useful for scene change detection
imshow(imdivide(I1c, I2));
```
